```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configura visualização
plt.style.use('seaborn-v0_8')
sns.set_palette("Set2")

# Caminho do arquivo CSV (coloque na pasta ../data)
caminho_csv = '../data/acidentes_exemplo.csv'

# Leitura do arquivo
df = pd.read_csv(caminho_csv, encoding='latin1', sep=';')

# Visualizar as 5 primeiras linhas
df.head()

# Ver estrutura do DataFrame
df.info()

# Ver estatísticas básicas
df.describe()

plt.figure(figsize=(12,6))
df['uf'].value_counts().sort_values(ascending=False).plot(kind='bar')
plt.title('Total de Acidentes por UF')
plt.xlabel('Estado')
plt.ylabel('Quantidade de acidentes')
plt.grid(True)
plt.show()

plt.figure(figsize=(12,6))
df['horario'].value_counts().sort_index().plot(kind='line', marker='o')
plt.title('Distribuição de Acidentes por Horário')
plt.xlabel('Hora do dia')
plt.ylabel('Quantidade')
plt.grid(True)
plt.show()

plt.figure(figsize=(10,5))
df['veiculo'].value_counts().plot(kind='barh')
plt.title('Tipos de veículos envolvidos')
plt.xlabel('Quantidade')
plt.grid(True)
plt.show()

print("Total de registros:", len(df))
print("Estados com mais acidentes:", df['uf'].value_counts().head(3).to_dict())
print("Horário mais crítico:", df['horario'].mode()[0])

```
